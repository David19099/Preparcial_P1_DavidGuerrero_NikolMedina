```python
# Solicita al usuario el código del cajero y lo convierte a entero
codigoCajero = int(input("Ingresa el código de cajero (###): "))

# Ciclo para validar que el tope de recaudo sea un valor válido
while True:
    try:
        # Solicita el valor máximo que puede recaudar el cajero
        topeRecaudo = int(input("Ingresa el valor máximo de reacudo: "))

        # Verifica que el valor ingresado sea mayor o igual a cero
        if topeRecaudo >= 0:
            break
        else:
            # Mensaje mostrado cuando se ingresa un número negativo
            print("El número debe ser mayor que cero. Inténtalo de nuevo.")

    # Captura el error cuando el usuario ingresa algo que no es un número
    except ValueError:
        print("Valor inválido. Inténtalo de nuevo")


# Inicializa el contador de transacciones en cero
contadorTransacciones = 0

# Inicializa el acumulador del dinero recaudado en cero
totalRecaudado = 0


# Ciclo principal que permite registrar las transacciones
while True:

    # Ciclo encargado de validar el monto recibido
    while True:
        try:
            # Solicita el monto recibido en la transacción
            montoRecibido = int(input("Ingrese el monto recibido: "))

            # Verifica que el monto sea mayor o igual a cero
            if montoRecibido >= 0:
                break
            else:
                # Mensaje mostrado si se ingresa un número negativo
                print("El número debe ser mayor que cero. Inténtalo de nuevo.")

        # Captura el error si se ingresa un valor que no sea numérico
        except ValueError:
            print("Valor inválido. Inténtalo de nuevo")


    # Suma el monto recibido al total recaudado
    totalRecaudado += montoRecibido

    # Aumenta en uno la cantidad de transacciones realizadas
    contadorTransacciones += 1


    # Verifica si el total recaudado alcanzó o superó el tope establecido
    if totalRecaudado >= topeRecaudo:

        # Informa que la caja debe suspenderse
        print("CAJA SUSPENDIDA: se alcanzó el tope de recaudo. Dirijase a tesorería")

        # Guarda el motivo por el cual terminó el turno
        motivoCierre = "Tope Alcanzado"

        # Finaliza el ciclo principal
        break


    # Verifica si el monto recibido es cero
    if montoRecibido == 0:

        # Guarda el motivo por el cual terminó el turno
        motivoCierre = "Fin de Cola"

        # Finaliza el ciclo principal
        break


# Calcula el promedio de dinero recibido por cada transacción
promedioTransacciones = totalRecaudado / contadorTransacciones


# Formatea el tope de recaudo agregando separadores de miles
# y reemplaza la coma por un apóstrofe
tope_formateado = f"{topeRecaudo:,}".replace(",", "'")

# Formatea el total recaudado
recaudo_formateado = f"{totalRecaudado:,}".replace(",", "'")

# Redondea el promedio a dos decimales y lo formatea
promedio_formateado = f"{round(promedioTransacciones, 2):,}".replace(",", "'")


# Muestra el reporte final del turno
print(f"""
----- REPORTE DE TURNO -----

# Muestra el código del cajero
Cajero:                 C-{codigoCajero}

# Muestra el tope de recaudo asignado
Tope asignado:          ${tope_formateado} COP

# Muestra la cantidad de transacciones realizadas
Transacciones:          {contadorTransacciones}

# Muestra el total de dinero recaudado
Recaudo total:          ${recaudo_formateado} COP

# Muestra el promedio recaudado por transacción
Promedio/transacción:   ${promedio_formateado} COP

# Muestra la razón por la que terminó el ciclo
Motivo de cierre:       {motivoCierre}
""")
```
