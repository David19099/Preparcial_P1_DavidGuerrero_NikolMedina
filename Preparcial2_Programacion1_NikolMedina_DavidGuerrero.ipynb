```python
codigoCajero = int(input("Ingresa el codigo de cajero"))
topeRecaudo = int(input("Ingresa el valor maximo de reacudo"))

contadorTransacciones = 0
totalRecaudado = 0
# montoRecibido=(input("Ingresa el monto a depositar"))
# try:
#     # Intentamos convertir la entrada a un número decimal (float)
#     recaudoRecibido1 = float(montoRecibido)

#     # Verificamos si es un número entero exacto (ej. 5.0 o 5)
#     if recaudoRecibido1.is_integer():
#         print(f"El número '{int(recaudoRecibido1)}' es un entero válido.")

# except ValueError:
#     # Si la conversión falla, no es un número
#     print("Error: No ingresaste un número válido.")


while True :
    montoRecibido = int(input("Ingrese el monto recibido: "))
    totalRecaudado += montoRecibido
    contadorTransacciones += 1
    # validacion adjuntarla
    if totalRecaudado >= topeRecaudo :
        print ("CAJA SUSPENDIDA: se alcanzó el tope de recaudo. Dirijase a tesorería")
        motivoCierre = "tope alcanzado"
        break
    if montoRecibido == 0 :
        motivoCierre = "fin de cola"
        break
print (" Codigo Caja: ", codigoCajero, " Máximo Recaudo; ", topeRecaudo, " Total recaudado en caja: ", totalRecaudado, " Transacciones realizadas correctamente: ", contadorTransacciones, " Motivo de cierre: ",  motivoCierre)
```
